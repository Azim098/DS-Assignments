```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("ABC Company.csv")
df.head()
```

Preprocessing

```python
df["Height"]=np.random.randint(150,181,len(df))
df.head()
```

Analysis Tasks

```python
team_count=df["Team"].value_counts()
team_percent=(team_count/len(df))*100
empteam=pd.DataFrame({"Employee Count":team_count,"Percentage":team_percent})
empteam
```

```python
pos_count=df["Position"].value_counts()
print(pos_count)
```

```python
bins=[18,25,30,35,40]
labels=['19-25','26-30','31-35','36-40']
df['Age Group']=pd.cut(df['Age'],bins=bins,labels=labels)
age_group=df['Age Group'].value_counts()
print(age_group)
```

```python
team_salary=df.groupby('Team')['Salary'].sum().sort_values(ascending=False)
team_salary.head(1)
```

```python
pos_salary=df.groupby('Position')['Salary'].sum().sort_values(ascending=False)
pos_salary.head(1)
```

```python
correlation=df['Age'].corr(df['Salary'])
print(correlation)
```

Graphical Representation

```python
team_count.plot(kind='bar',figsize=(13,6))
plt.title("Employees in each team")
plt.xlabel("Team")
plt.ylabel("Count")
plt.show()
```

```python
pos_count.plot(kind='pie',autopct='%1.1f%%')
plt.title("Employee Positions")
plt.ylabel("")
plt.show()
```

```python
age_group.plot(kind='bar')
plt.title("Age Group Distribution")
plt.xlabel("Age group")
plt.ylabel("Employees")
plt.show()
```

```python
team_salary.plot(kind='bar', figsize=(12,5))
plt.title("Salary Expenditure by Team")
plt.show()
```

```python
pos_salary.plot(kind='bar')
plt.title("Salary Expenditure by Position")
plt.show()
```

```python
plt.scatter(df['Age'], df['Salary'])
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```
