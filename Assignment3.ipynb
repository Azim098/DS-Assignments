\Visualization with Seaborn and Matplotlib

Use the “Titanic” dataset from Seaborn and do the following:
1. Countplot: Passenger Count by Embarked Port
● Create a countplot to display the number of passengers who boarded from each
port (embarked), separated by sex.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style("whitegrid")
iris = sns.load_dataset("titanic")
iris.head()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=iris,x='embarked',hue='sex')
plt.title("Passenger Count by Embarked Port")
plt.xlabel("Embarked Port")
plt.ylabel("No Of Passenger")
plt.show()
print("Observation:")
print("- Most passengers boarded from S.")
print("- C has fewer passengers than S.")
print("- Male passengers are more than female passengers at every port.\n")
```

2. Swarmplot: Age Distribution across Classes
● Create a swarmplot to visualize the distribution of age across passenger classes
(pclass), categorized by sex.

```python
plt.figure(figsize=(10,6))
sns.swarmplot(data=iris,x='pclass',y='age',hue='sex',dodge=True)
plt.title("Age Distribution across Classes")
plt.xlabel("P class")
plt.ylabel("Age")
plt.legend(title="Sex")
plt.show()
print("Observation:")
print("- First-class passengers include many adults and elderly people.")
print("- Third-class passengers have a wider spread of ages.")
print("- Children are present in all passenger classes.")
```

3. Pie Chart: Gender Distribution
● Create a pie chart using Matplotlib to show the percentage of male vs. female
passengers. - Add percentage labels using autopct - Use explode for one slice to
highlight it.

```python
gender_count = iris['sex'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(gender_count,labels=gender_count.index,autopct='%1.1f%%',explode=(0.1,0))

plt.title("Gender Distribution of Passengers")
plt.show()
print("Observation:")
print("- Male passengers make up the majority of the passengers.")
print("- Female passengers represent a smaller proportion.")
```

4. Heatmap: Correlation Matrix
● Create a heatmap to show correlations between age, fare, pclass, sibsp, parch,
and survived. (Use seaborn.heatmap with annotations)

```python
plt.figure(figsize=(8,6))
corr=iris[['age','fare','pclass','sibsp','parch','survived']].corr()
sns.heatmap(corr,annot=True)
plt.title("Correlation Matrix")
plt.show()
print("Observation:")
print("- Passenger class is negatively correlated with fare.")
print("- Fare has a slight positive correlation with survival.")
```

5. Violin Plot: Fare by Class and Gender
● Create a violin plot to visualize fare distribution across classes, separated by
gender.

```python
plt.figure(figsize=(9,6))
sns.violinplot(data=iris,x='pclass',y='fare',hue='sex')
plt.title("Fare Distribution by Class and Gender")
plt.xlabel("P Class")
plt.ylabel("Fare")
plt.legend(title="Sex")
plt.show()
print("Observation:")
print("- First-class passengers paid the highest fares.")
```

6. Scatter Plot: Age vs Fare
● Create a scatter plot to explore the relationship between age and fare, colored by
survival status.
● Use a seaborn scatterplot with hue='survived'. Each plot must have titles, labels,
and legends where needed.

```python
plt.figure(figsize=(9,6))
sns.scatterplot(data=iris,x='age',y='fare',hue='survived')
plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived")
plt.show()
print("Observation:")
print("- Higher fare passengers had a better chance of survival.")
print("- Most passengers paid low fares.")
```
